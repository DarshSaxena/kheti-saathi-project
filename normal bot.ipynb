```python
import random

def chatbot():
    print(" Kheti Saathi AI Chatbot (type 'exit' to stop)\n")

    while True:
        query = input("You: ")
        if query.lower()=="hi":
            print("Chatbot: How can I help you")
            continue

        if query.lower() == "exit":
            print("Chatbot: Goodbye ")
            break

        response = get_response(query)
        print("Chatbot:", response)


def get_response(query):
    q = query.lower()

    if "crop" in q or "grow" in q:
        responses = [
            " Based on your conditions, rice would be a suitable crop due to high rainfall compatibility.",
            " You can consider growing paddy as it performs well in moist conditions.",
            " Rice is a recommended crop considering water availability and soil type."
        ]
        return random.choice(responses)

    elif "disease" in q or "leaf" in q or "spots" in q:
        responses = [
            " This seems like a fungal infection. Try using a fungicide spray.",
            " Leaf spots may indicate disease. Remove affected leaves and apply treatment.",
            " It could be a plant disease. Proper diagnosis via image is recommended."
        ]
        return random.choice(responses)

    elif "fertility" in q or "yield" in q:
        responses = [
            " To improve soil fertility, use organic compost and crop rotation.",
            " Adding manure and maintaining proper irrigation can increase yield.",
            " Balanced fertilizers and good soil management improve productivity."
        ]
        return random.choice(responses)

    elif "water" in q or "irrigation" in q:
        responses = [
            " Use drip irrigation to save water and improve efficiency.",
            " Avoid overwatering as it can damage crops.",
            " Proper irrigation scheduling is important for healthy crops."
        ]
        return random.choice(responses)

    else:
        responses = [
            " Based on your query, proper farming practices and monitoring are recommended.",
            " It depends on environmental conditions. Can you provide more details?",
            " Smart farming techniques can help improve your results."
        ]
        return random.choice(responses)


chatbot()
```
